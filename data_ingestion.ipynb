```python
import pandas as pd
import time
from tqdm import tqdm

# Create iterator
df_iter = pd.read_csv(
    prefix + 'yellow_tripdata_2021-01.csv.gz',
    dtype=dtype,
    parse_dates=parse_dates,
    chunksize=10000
)

# Insert chunks
for i, df_chunk in enumerate(tqdm(df_iter)):

    if i == 0:
        df_chunk.to_sql(
            name="yellow_taxi_data",
            con=engine,
            if_exists="replace",
            index=False
        )
    else:
        df_chunk.to_sql(
            name="yellow_taxi_data",
            con=engine,
            if_exists="append",
            index=False
        )

    # end = time.time()

    # print(f"Finished chunk {i}")
    # print(f"Inserted rows: {len(df_chunk)}")
    # print(f"Time: {end - start:.2f} sec")
    
```
